What is this graph showing?

“We are checking whether students' average scores are different depending on their lunch category.”

There are two lunch categories:

Standard
Free/Reduced

The x-axis represents the student's average score.

The y-axis represents how many students have scores in that range.

The bars show the actual distribution of students' scores, while the smooth line (KDE) shows the general shape of the distribution.

1. First graph — All students

The first graph compares all students.

We can see that:

Students with standard lunch generally have higher average scores.
Their scores are mostly concentrated around 65–80.
Students with free/reduced lunch generally have lower average scores.
Their scores are more concentrated around 55–70.
Simple explanation:

“Students with standard lunch generally have higher average scores than students with free or reduced lunch.”

2. Second graph — Female students

The second graph looks only at female students.

Again, we can see:

Standard lunch students tend to have higher average scores.
Free/reduced lunch students tend to have lower average scores.
The two distributions overlap, meaning this is not true for every individual student.
Simple explanation:

“Among female students, the students with standard lunch generally have higher average scores than those with free or reduced lunch.”

3. Third graph — Male students

The third graph looks only at male students.

We see a similar pattern:

Standard lunch students are mostly concentrated around higher scores.
Free/reduced lunch students are more concentrated around lower scores.
There is still some overlap between the two groups.
Simple explanation:

“Among male students, standard-lunch students also generally have higher average scores than free/reduced-lunch students.”

4.3 Multivariate analysis using pieplot

```python
df['gender'].value_counts()
```

```python
plt.rcParams['figure.figsize'] = (30, 12)

plt.subplot(1, 5, 1)
size = df['gender'].value_counts()
labels = 'Female', 'Male'
color = ['red','green']


plt.pie(size, colors = color, labels = labels,autopct = '.%2f%%')
plt.title('Gender', fontsize = 20)
plt.axis('off')



plt.subplot(1, 5, 2)
size = df['race_ethnicity'].value_counts()
labels = 'Group C', 'Group D','Group B','Group E','Group A'
color = ['red', 'green', 'blue', 'cyan','orange']

plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Race_Ethnicity', fontsize = 20)
plt.axis('off')



plt.subplot(1, 5, 3)
size = df['lunch'].value_counts()
labels = 'Standard', 'Free'
color = ['red','green']

plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Lunch', fontsize = 20)
plt.axis('off')


plt.subplot(1, 5, 4)
size = df['test_preparation_course'].value_counts()
labels = 'None', 'Completed'
color = ['red','green']

plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Test Course', fontsize = 20)
plt.axis('off')


plt.subplot(1, 5, 5)
size = df['parental_level_of_education'].value_counts()
labels = 'Some College', "Associate's Degree",'High School','Some High School',"Bachelor's Degree","Master's Degree"
color = ['red', 'green', 'blue', 'cyan','orange','grey']

plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Parental_Education', fontsize = 20)
plt.axis('off')


plt.tight_layout()
plt.grid()

plt.show()
```

#####  Insights
- Number of Male and Female students is almost equal
- Number students are greatest in Group C
- Number of students who have standard lunch are greater
- Number of students who have not enrolled in any test preparation course is greater
- Number of students whose parental education is "Some College" is greater followed closely by "Associate's Degree"
